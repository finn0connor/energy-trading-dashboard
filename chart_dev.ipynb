```python
import pandas as pd
import plotly.graph_objects as go

from data.semo_data.semo import get_wind_forecast

# Get the data
df = get_wind_forecast("2026-01-01", "2026-03-27")

# Plot using StartTime column
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df['StartTime'], 
    y=df['Forecast'], 
    mode='lines', 
    name='All-Ireland Wind Forecast'
))
fig.show()
```

```python
from data.semo_data.semo import get_daily_load_forecast

# Get the data
df = get_daily_load_forecast("2026-01-01", "2026-03-27")

# This is the systemwide demand forecast
# Plot using StartTime column
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df['StartTime'], 
    y=df['AggregatedForecast'], 
    mode='lines', 
    name='Total System Load Forecast'
))
fig.show()
```
